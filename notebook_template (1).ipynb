# Mark Down

Bold/Italic: Use **text** for bold and *text* for italics. You can combine them ***like this*** for both Markdown Guide.

Links: The syntax is [Link Text](URL). This is perfect for citing your data sources Daring Fireball.

Inline Code: Use single backticks ` to highlight variable names or small snippets like `np.array()`.

Images: You can even put a small image or icon inside a cell using ![Alt Text](ImageURL).

# Esc - command mode, M - markdown Y - Code
A :	    Insert cell Above

B :	    Insert cell Below

D, D :	Delete current cell (press D twice)

Z :	    Undo cell deletion

Shift + V :	Paste cell above

L :	    Toggle Line numbers (great for debugging)

Shift + Enter :	Run cell and move to the next one

Ctrl + Enter :	Run cell and stay in the same spot

Alt + Enter : Run cell and insert a new one below

1-6 Heading 1 - 6

Unordered Lists: Use *, -, or +.

Ordered Lists: Use 1., 2., etc.

Nested Lists: Indent by four spaces Markdown Guide.

Tables are built using pipes | and dashes -. You don't need to make the spaces line up perfectly; Jupyter will fix the alignment for you when you run the cell Daring Fireball.  Use three dash (---)

| :--- | = Left Aligned

| :---: | = Centered

| ---: | = Right Aligned

To Aligh table to the left, use the html code

%%html
<style>
    table { margin-left:0 !important; float: left; clear:both }
</style>
