# Mark Down

Bold/Italic: Use **text** for bold and *text* for italics. You can combine them ***like this*** for both Markdown Guide.

Links: The syntax is [Link Text](URL). This is perfect for citing your data sources Daring Fireball.

Inline Code: Use single backticks ` to highlight variable names or small snippets like `np.array()`.

Images: You can even put a small image or icon inside a cell using ![Alt Text](ImageURL).

# Esc - command mode, M - markdown Y - Code
A :	    Insert cell Above

B :	    Insert cell Below

D, D :	Delete current cell (press D twice)

Z :	    Undo cell deletion

Shift + V :	Paste cell above

L :	    Toggle Line numbers (great for debugging)

Shift + Enter :	Run cell and move to the next one

Ctrl + Enter :	Run cell and stay in the same spot

Alt + Enter : Run cell and insert a new one below

1-6 Heading 1 - 6

Unordered Lists: Use *, -, or +.

Ordered Lists: Use 1., 2., etc.

Nested Lists: Indent by four spaces Markdown Guide.

Tables are built using pipes | and dashes -. You don't need to make the spaces line up perfectly; Jupyter will fix the alignment for you when you run the cell Daring Fireball.  Use three dash (---)

| :--- | = Left Aligned

| :---: | = Centered

| ---: | = Right Aligned

To Aligh table to the left, use the html code

%%html
<style>
    table { margin-left:0 !important; float: left; clear:both }
</style>

# 🚀 Data Science Speed Test: Python vs. Numpy
---
### **Project Overview**
This notebook documents the performance difference between standard Python and NumPy using **IPython Magic Commands**.

| Tool | Category | Documentation |
| :--- | :---: | ---: |
| **Python** | *Base Language* | [Python Link](https://docs.python.org) |
| **NumPy** | *Math Library* | [Numpy Link](https://numpy.org) |
| **Matplotlib** | *Visualization* | [MatPlotLib Link](https://matplotlib.org) |


<div style="clear: both;"></div>

> **Goal:** Prove that vectorization is superior for large-scale data processing.


```python
# Using cell instead of markdown to display html page

from IPython.display import display, HTML

display(HTML('<a href="https://pandas.pydata.org" target="_blank">Click here to see Pandas Docs</a>'))
```

```python
import numpy as np

#1. Prepare a loarge dataset
data_size = 1000
data_array = np.arange(data_size)

print(f"Testing math on {data_size:,} elements...")

#2. Time the stardard Python list
print(f"Python List Comprehension Speed:")
%timeit [x**2 for x in range(data_size)]

#3. Time the NumPy way
print(f"NumPy Vectorized Speed:")
%timeit data_array ** 2

```
