```python
import numpy as np
```

**Masked Array**

```python
array1 = np.array([1,5,np.nan, 53, 2, np.nan])

# create and display a mask and check for nan item 
mask = np.isnan(array1)
print(mask)
# create and display a mask array to mask out all nans
masked_array = np.ma.array(array1,mask=mask)
print(masked_array)

# another example
array2 = np.arange(1,11)
mask1 = np.array([num%2==0 for num in array2])
print(array2)
print(mask1)
masked_array1 = np.ma.array(array2,mask=mask1)
print(masked_array1)

filled_array = np.ma.filled(masked_array1, fill_value=-1)
print(filled_array)
```

```python
array1 = np.arange(1,11)

# create and display a mask to mask for all even numbers
array2 = np.arange(1,11)
mask1 = np.array([num%2==0 for num in array2])
print(array2)
print(mask1)

# create and display a masked array using the mask
masked_array1 = np.ma.array(array2,mask=mask1)
print(masked_array1)

# create and print a filled array to fill the mask with -1 using fill_value argument
filled_array = np.ma.filled(masked_array1, fill_value=-1)
print(filled_array)
```

**Structured array**

```python
# structured array allows you to have different data types existed in an array, using dtype to create the structure
# structed array is one D array

# using a list of tuples, each tuple represent one field with name and data type
dtype = [('id', 'i4'),('name','U15'),('weekly_pay','f4')]

# using np.dtype to directly set the data type for different fields
dtype1 = np.dtype('i4, U15, f4')

# using np.type() to create a dictionary with 2 keys: names and formats, names key contains a list of attribute names,
#   and formats key contains a list of data type for each attribute

dtype2 = np.dtype({
    'names':['id','name','weekly_pay'],
    'formats':['i4','U15','f4' ]})

# create a data array using list of tuples
emp_data = np.array([
    (1, 'JSmith',1500.0),
    (2, 'DCook', 1500.0),
    (3, 'AJones',4000)],dtype = dtype)

# display data array and its data type
emp_data
```

```python
# display the first employee's pay, access row and field

print(emp_data[0]['weekly_pay'])
# change JSmith pay to 3000
emp_data[0]['weekly_pay'] = 3000
emp_data[0]
```

```python
# add new employee

# create a new employee
new_emp = np.array([(4, 'JMark',7500.0)],dtype=dtype)
# append the new employee to emp_data
emp_data = np.append(emp_data, new_emp)
emp_data
```

```python
# delete employee - (s_array,index,axis)

emp_data = np.delete(emp_data,1,0)
```

```python
# filter data
# high pay - show pay > 5000

high_pay = emp_data['weekly_pay'] > 5000
emp_data[high_pay]
```

```python
emp_data[0]['weekly_pay'] = 1500
emp_data
```

```python
# sort data - use [::-1] for decending 

# sort pay - default ascending
sorted_pay = np.sort(emp_data, order=['weekly_pay','name'])
sorted_pay1 = np.sort(emp_data['weekly_pay'])
print(sorted_pay)
sorted_pay1
# different sort order by fields
name_descending = np.argsort(emp_data['name'])[::-1]
pay_ascending = np.argsort(emp_data['weekly_pay'])
sort_indices = np.lexsort((name_descending, pay_ascending))
sorted_data = emp_data[sort_indices]
sorted_data
```

```python
# record array
rec_array = emp_data.view(np.recarray)
print(rec_array)
```

```python
print(rec_array['name'])
print(rec_array.name)
print(np.sum(rec_array.weekly_pay > 2000))
print(rec_array[rec_array.weekly_pay>2000])
```

**Ufunc**

```python
# ufunc
def prod(a,b):
    return a * b

test = np.frompyfunc(prod,2,1)
print(test([1,2,3],[4,5,6]))

# check type
type(test)
type(np.add)
```

**Other functions, reduce, accumulate, argmin, argmax, nsnsum, any, all, partition, unique**

```python
array1 = np.arange(1,11) # use add.reduce, apply multiply.reduce

result = np.add.reduce(array1)
print(result)
```

```python
# use add.accumulate
print(array1)
result = np.add.accumulate(array1)
print(result)
```

```python
# nan - not a number, use ~np.isnan

array1 = np.array([1,np.nan, 56, 23,np.nan, 6, np.nan,76])

# create nan mask and not_nan mask
print(array1)
nan = np.isnan(array1)
not_nan = ~np.isnan(array1)
print(nan, not_nan)
print(array1[nan])
print(array1[not_nan])

# argmax - index of largest value, argmin - index of lowest value, nanargmax - index of largest value ignored nan
print(np.argmax(array1)) # show max index including nan - 1
print(np.nanargmax(array1)) # max index ignore nan - 7

print(np.nansum(array1)) # sum all elements ignore nan
print(np.sum(array1)) # with nan - result is nan

print(any(array1>5)) # True if any number is > 5
print(all(array1>5)) # True only all numbers are > 5
print(array1[array1>5])

# count unique
array = np.array([1,1,2,3,3,3,5,6])
unique_value, counts = np.unique(array, return_counts=True)
print(unique_value, counts)
```

```python
# partition
array1 = np.array([7,3,2,1,6,5,4])
print(array1)
# the element at index 4 moves to its absolute correct sorted position
partitioned_array = np.partition(array1,2)
print(partitioned_array)

# 2d
rng = np.random.default_rng(123)
array_2d = rng.integers(10,50,size=(3,3))
print(f'Original array:\n{array_2d}')
p = np.partition(array_2d, 1,axis=0)
print(f'partition across the row:\n{p}')
p1 =np.partition(array_2d,1,axis=1)
print(f'partition across the column:\n{p1}')
```

```python
# Example of ndarray broadcasting

# one D
array1 = np.array([1,2,3])
# two D
array2 = np.arange(1,7).reshape(2,3)

print(array1, array1.shape,'\n\n',array2, array2.shape)
array1 + array2

# add dimension 
array1_row = array1[np.newaxis,:]
array1_col = array1[:, np.newaxis]
print(f'\nadd dim by row:\n{array1_row} shape: {array1_row.shape}')
print(f'add dim by col:\n{array1_col} shape: {array1_col.shape}')

# check on math operation for 2 arrays with the same dimension
print(array2 + array1_row) # add (1,3) and (2,3)
print(array2 + array1_col) # caue error cause cannot add (2,3), (3,1)
```
