## Use Numpy

##### NumPy (Numerical Python) is a fundamental package for scientific computing in Python. <br>It provides support for large, multi-dimensional arrays and matrices, along with a collection of <br>high-level mathematical functions to operate on these arrays efficiently.

```python
# import numpy 
import numpy as np

# display function
def display(*args):
    i = 0
    for arg in args:
        print(f'{arg}' if i%2==0 else f'{arg}\n')
        i+=1
```

#### Create ndarray

```python
# 1. Create ndarray using python list : [1,2,3,4,5]

array1 = np.array([1, 2, 3, 4, 5])
display("NumPy array:", array1)
type(array1)
```

```python
# 2. Create a two-d ndarray. It conains 2 rows and 3 columns

array2 = np.array([[1, 2, 3],
                   [4, 5, 6]])
display("2D array:", array2)

type(array2)
```

```python
# 3. Create a ndarray using arange() 1-5 with data type specified
array3 = np.arange(1, 6, dtype=int)
display("Array using arange:", array3)
```

```python
# 4. Create a ndarray with 5 evenly spaced numbers between 0 and 1

array4 = np.linspace(0, 1, 5)
display("Evenly spaced array:", array4)
```

```python
# 5. Create a two-d ndarray using np.diag() - 1,2,3,4 in diagonal position

array5 = np.diag([1, 2, 3, 4])
array6 = np.arange(1, 10).reshape(3, 3)
display("Diagonal array:", array5, "3 × 3 array:", array6)
np.diag(array6, k=-1)
```

```python
# 6. Create a two-d ndarray with ones on the diagonal line and zeros elsewhere using np.eye()

array6 = np.eye(3)
print(array6)
```

```python
# 7. Create a two-d ndarray with 3 by 3 zeros

array7 = np.zeros((3, 3))
print(array7)
```

```python
# 8. Create a two-d ndarray with 3 by 3 ones

array8 = np.ones((3, 3))
print(array8)
```

```python
# 9. Create a two-d ndarray with 3 by 3 filled with 7 using np.full()

array9 = np.full((3, 3), 7)
display(array9)
```

```python
# 10. Create a ndarray using np.asarray - use list or tuple

array10 = np.asarray([1, 2, 3, 4, 5])
print(array10)
```

```python
# 11. Explore ndarray attributes such as shape, dtype, ndim, size, itemsize, nbytes, etc.
array1= np.arange(1,11)
array2 = np.arange(1,11).reshape(2,5)

array1 = np.arange(1, 11)
array2 = np.arange(1, 11).reshape(2, 5)

for arr in (array1, array2):
    print(arr)
    print("Shape:", arr.shape)        # Length along each dimension
    print("Data type:", arr.dtype)    # Type of stored elements
    print("Dimensions:", arr.ndim)    # Number of dimensions
    print("Size:", arr.size)          # Total number of elements
    print("Item size:", arr.itemsize) # Bytes per element
    print("Total bytes:", arr.nbytes) # Bytes used by element data
    print()
```

#### Array Broadcasting

```python
rng = np.random.default_rng(seed=123)
matrix = rng.integers(10,50,(3,3))
print(matrix)
print(f"---------------------------")
one_row_matrix = matrix.reshape(1,-1)
one_col_matrix = matrix.reshape(-1,1)
print(f"---------------------------")
display(one_row_matrix, one_col_matrix)
print(f"---------------------------")

combined = one_row_matrix + one_col_matrix
display(one_row_matrix.shape, one_col_matrix.shape,combined)
```

#### Iterating over arrays

```python
array1 = np.array([1,2,3])
array2 = np.arange(12).reshape(3,4)

for num in array1:
    print(num, end = " ")
print()

#2d

for row in array2:
    for col in row:
        print(col, end = " ")
    print()
print()

#2d using index

for i in range(array2.shape[0]):
    for j in range(array2.shape[1]):
        print(array2[i,j], end = " ")
    print()
```

#### Using concatenate, vstack and hstack

```python
# concatenate 
arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

print(np.concatenate((arr1,arr2), axis=0))
print(np.concatenate((arr1,arr2), axis=1))
```

```python
# vstack, hstack

print(np.vstack((arr1, arr2)))
print(np.hstack((arr1, arr2)))
```
