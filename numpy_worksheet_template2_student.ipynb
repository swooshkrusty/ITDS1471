```python
# nan - not a number, use ~np.isnan

array1 = np.array([1,np.nan, 56, 23,np.nan, 6, np.nan,76])

# create nan mask and not_nan mask



# argmax - index of largest value, argmin - index of lowest value, nanargmax - index of largest value ignored nan
print(np.argmax(array1)) # show max index including nan - 1
print(np.nanargmax(array1)) # max index ignore nan - 7

print(np.nansum(array1)) # sum all elements ignore nan
print(np.sum(array1)) # with nan - result is nan

print(any(array1>5)) # True if any number is > 5
print(all(array1>5)) # True only all numbers are > 5
print(array1[array1>5])

# count unique
array = np.array([1,1,2,3,3,3,5,6])
unique_value, counts = np.unique(array, return_counts=True)
print(unique_value, counts)
```

```python
# partition
array1 = np.array([7,3,2,1,6,5,4])

# the element at index 4 moves to its absolute correct sorted position
partitioned_array = np.partition(array1,2)
print(partitioned_array)

# 2d
rng = np.random.default_rng(123)
array_2d = rng.integers(10,50,size=(3,3))
print(f'Original array:\n{array_2d}')
p = np.partition(array_2d, 1,axis=0)
print(f'partition across the row:\n{p}')
p1 =np.partition(array_2d,1,axis=1)
print(f'partition across the column:\n{p1}')
```

```python
# Example of ndarray broadcasting

# one D
array1 = np.array([1,2,3])
# two D
array2 = np.arange(1,7).reshape(2,3)

print(array1, array1.shape,'\n\n',array2, array2.shape)
array1 + array2

# add dimension 
array1_row = array1[np.newaxis,:]
array1_col = array1[:, np.newaxis]
print(f'\nadd dim by row:\n{array1_row} shape: {array1_row.shape}')
print(f'add dim by col:\n{array1_col} shape: {array1_col.shape}')

# check on math operation for 2 arrays with the same dimension
print(array2 + array1_row) # add (1,3) and (2,3)
print(array2 + array1_col) # caue error cause cannot add (2,3), (3,1)
```
